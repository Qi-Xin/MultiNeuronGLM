```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets
from torchvision.transforms import ToTensor
from numpy import random
```

```python
# Generate data and Dataloader object

class MyDataset(Dataset):
  def __init__(self, Ndata, Ndim):
    x = random.uniform(low=-1.0, high=1.0, size=(Ndata,Ndim))
    y = random.randint(2, size=(Ndata))

    self.x_train = torch.tensor(x, dtype=torch.float32)
    self.y_train = torch.tensor(y, dtype=torch.float32)

  def __len__(self):
    return len(self.y_train)
  
  def __getitem__(self,idx):
    return self.x_train[idx], self.y_train[idx]
```

```python
class NeuralNetwork(nn.Module):

    def __init__(self, Ndim, Nsyn):
        super(NeuralNetwork, self).__init__()
        self.linear = nn.Linear(Ndim, 1, bias=False)
        self.final_weight = -nn.Parameter(torch.Tensor([1.0]))
        self.final_bias = -nn.Parameter(torch.Tensor([0.5]))

    def forward(self, x):
        pred = self.linear(x)*self.final_weight + self.final_bias
        return pred.squeeze()

# class NeuralNetwork(nn.Module):

#     def __init__(self, Ndim, Nsyn):
#         super(NeuralNetwork, self).__init__()
        
# #         self.weights = nn.Parameter(1/Nsyn/Ndim * torch.ones(1, Ndim, Nsyn))
#         self.weights = nn.Parameter(10 * torch.ones(1, Ndim, Nsyn))
#         self.positions = nn.Parameter(torch.rand(1, Ndim, Nsyn))
#         self.log_amplitudes = nn.Parameter(np.log(1/Nsyn/Ndim) * torch.ones(1, Ndim, Nsyn))
#         self.syn_sigmoid = nn.Sigmoid()
#         self.final_weight = -nn.Parameter(torch.Tensor([1.0]))
#         self.final_bias = -nn.Parameter(torch.Tensor([0.5]))

#     def forward(self, x):
#         x = x[:, :, None]
#         link_scale = torch.mul(self.weights, x-self.positions)
        
#         response_scale = self.syn_sigmoid(link_scale)
#         amplitudes = torch.exp(self.log_amplitudes)
#         with_amp = torch.mul(response_scale, amplitudes)
#         pred = torch.sum(with_amp, axis=(1,2) )*self.final_weight + self.final_bias
#         return pred.squeeze()
```

```python
def train(dataloader, model, loss_fn, optimizer):
    size = len(dataloader.dataset)
    model.train()
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)
        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)

        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        loss = loss.item()
#     print(f"train loss: {loss}")
    return loss

def test(dataloader, model, verbose=False):
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    model.eval()
    test_loss = 0.0
    with torch.no_grad():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)
            pred = model(X)
            binary_pred = (pred>=0)
            binary_y = (y>=0.5)
            test_loss += torch.sum(binary_pred!=binary_y)
    test_loss /= y.shape[0]
#     print(f"Percent wrong: : {test_loss:>3f} \n")
    if verbose:
        print(pred)
        print(y)
    return test_loss
```

```python
def myReLuLoss(pred, labels):
    pm_labels = (2.0 * labels.float() - 1.0)
    return  (F.relu(- pm_labels * pred + 0.1 ) ).mean()
```

```python
# Tuning params
Ndim = 4
Nsyn = 1
Ndata = 4

# Fixed params
batch_size = Ndata
device = "cpu"

training_data = MyDataset(Ndata, Ndim)
train_dataloader = DataLoader(training_data, batch_size=batch_size)

epochs = 100
test_mse_rcd = np.zeros(epochs)

model = NeuralNetwork(Ndim, Nsyn)
loss_fn = nn.BCEWithLogitsLoss()
# loss_fn = myReLuLoss
optimizer = torch.optim.SGD(model.parameters(), lr=1e-1, momentum=0.9)
for t in range(epochs):
    loss = train(train_dataloader, model, loss_fn, optimizer)
    wrong = test(train_dataloader, model)
    if t % 10 == 0:
        print(f"loss: {loss:.5f}; Incorrect, {wrong:.3f}")
#         print(model.weights[0,0])
test(train_dataloader, model, verbose=True)
print("Done!")
```
